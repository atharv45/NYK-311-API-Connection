#Block 1: Cleaning

```python
df_latest[['created_date','closed_date']].dtypes
```

```python
df_latest['closed_date'].isna().sum()
```

```python
df_latest['borough'].value_counts()
```

```python
df_latest['borough'] = df_latest['borough'].str.upper()
df_latest['agency'] = df_latest['agency'].str.upper()
df_latest['complaint_type'] = df_latest['complaint_type'].str.upper()

df_latest['borough'] = df_latest['borough'].str.strip()
df_latest['agency'] = df_latest['agency'].str.strip()
df_latest['complaint_type'] = df_latest['complaint_type'].str.strip()
```

```python
df_latest['borough'].value_counts(dropna=False)
```

```python
#Create a days to close column
```

```python
df_latest['days_to_close'] = (df_latest['closed_date'] - df_latest['created_date']).dt.total_seconds()/ 86400
```

```python
df_latest['days_to_close'].describe()
```

#Task 1: Handle the negative values

```python
import numpy as np
```

```python
neg = df_latest['days_to_close'] < 0
print('Negative rows', neg.sum())
```

```python
df_latest.loc[neg,'days_to_close'] = np.nan
```

```python
print(df_latest['days_to_close'].min())
```
